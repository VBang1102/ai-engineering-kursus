3. Loop mod vektorisering. Beregn summen af kvadraterne af 1.000.000 tilfældige tal, først med en for løkke og derefter med NumPy. Tag tid på begge med time eller %timeit, og skriv hvor mange gange hurtigere NumPy er.

```python
import numpy as np
import time
summen_loop=0
summen_np=0

rng=np.random.default_rng(42)
x=rng.random(1000000)
x_liste=x.tolist()

start_loop=time.perf_counter()
for i in x_liste:
    summen_loop += i**2
slut_loop=time.perf_counter()
print(summen_loop)
tid_loop=slut_loop-start_loop
print(tid_loop)


start_np=time.perf_counter()
summen_np=np.sum(x**2)
slut_np=time.perf_counter()
print(summen_np)
tid_np=slut_np-start_np
print(tid_np)

print(np.isclose(summen_loop, summen_np))

faktor=tid_loop/tid_np
print(faktor)
```

Numpy er altså 57,6 gange hurtigere end python

4. Matrixprodukt i hånden. Skriv en funktion med indlejrede løkker, der ganger en 2 x 3 matrix med en 3 x 2 matrix. Tjek resultatet mod A @ B. Forklar i en markdown celle, hvorfor formen på resultatet bliver 2 x 2.

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[7, 8], [9, 10], [11, 12]])
print(a)
print(b)

matrixprodukt_np= np.matmul(a,b)
print(matrixprodukt_np)

def matrixprodukt(a, b):
    c = np.zeros((a.shape[0], b.shape[1]))
    for i in range(a.shape[0]):
        for j in range(b.shape[1]):
            for k in range(a.shape[1]):
                c[i, j] += a[i, k] * b[k, j]
    return c

matrixprodukt_loop=matrixprodukt(a,b)
print(matrixprodukt_loop)

```

Grunden til at det bliver en 2x2 matrix er at man tager produktet af række 1 i a med kolonne 1 i b for at få punkt (1,1). Dette gør man også med række 2 og kolonne 1 osv. Dermed ender man op med kun at have en 2X2 matrix.
